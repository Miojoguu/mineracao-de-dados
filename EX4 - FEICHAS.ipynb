### Exercício 1: Distância em passos de desvio-padrão

```python
media = 100
desvio = 5
valor = 115

distancia = valor - media
print(f"Distância entre o valor e a média: {distancia}")

desvios_na_distancia = distancia / desvio
print(f"Quantos desvios-padrão cabem nessa distância: {desvios_na_distancia}")

z_score = (valor - media) / desvio
print(f"Z-Score: {z_score}")
```

**Interpretação:** O valor 115 está 3 desvios-padrão acima da média do seu conjunto.

### Exercício 2: Acima ou abaixo da média?

```python
casos = [85, 100, 120]
media = 100
desvio = 10

for valor in casos:
    z_score = (valor - media) / desvio
    if z_score < 0:
        posicao = "abaixo da média"
    elif z_score == 0:
        posicao = "exatamente na média"
    else:
        posicao = "acima da média"
        
    print(f"Valor: {valor} | Z-Score: {z_score} | Posição: {posicao}")
```

**Interpretação dos Sinais:**
- **Sinal Positivo:** Indica que o valor é maior que a média.
- **Sinal Negativo:** Indica que o valor é menor que a média.
- **Zero:** Indica que o valor é exatamente igual à média.

### Exercício 3: Temperaturas

```python
temperaturas = [21, 23, 25, 27, 29]
media = 25
desvio = 2

z_scores = [(t, (t - media) / desvio) for t in temperaturas]

print("Z-Scores das temperaturas:")
for t, z in z_scores:
    print(f"{t}°C -> Z-Score: {z}")

# Encontrando o maior valor absoluto de Z
leitura_incomum = max(z_scores, key=lambda item: abs(item[1]))
print(f"\nA leitura mais distante da média (maior valor absoluto de Z) é {leitura_incomum[0]}°C com Z-Score absoluto de {abs(leitura_incomum[1])}")

# Como 21 e 29 tem a mesma distância, o max retorna o primeiro ou o último dependendo da implementação, 
# vamos mostrar todos que empatam na distância máxima:
max_dist = max([abs(z) for t, z in z_scores])
mais_distantes = [t for t, z in z_scores if abs(z) == max_dist]
print(f"Valores mais distantes em termos de desvio-padrão: {mais_distantes}")
```

### Exercício 4: Latência de uma API

```python
import numpy as np

latencias = [98, 102, 101, 99, 100, 103, 97, 180]

media = np.mean(latencias)
desvio = np.std(latencias)

print(f"Média: {media}")
print(f"Desvio-Padrão: {desvio}")

z_score_180 = (180 - media) / desvio
print(f"Z-Score da latência 180 ms: {z_score_180}")

if abs(z_score_180) > 3:
    print("\nO valor 180 ms MERECE INVESTIGAÇÃO (|Z| > 3).")
else:
    print("\nO valor não atinge o limite para investigação.")
```

**Lembrete de Segurança/Análise de Dados:** Um Z-Score alto indica um comportamento fora do padrão, exigindo que o evento seja investigado. No entanto, investigar **não significa apagar o dado automaticamente**, pois outliers podem revelar informações valiosas, como anomalias reais, gargalos ou ataques à API.

### Exercício 5: Monitoramento de CPU com classificação

```python
import numpy as np

cpu = [42, 45, 47, 44, 46, 43, 48, 92]

media = np.mean(cpu)
desvio = np.std(cpu)

print(f"Média: {media:.2f} | Desvio: {desvio:.2f}\n")

for valor in cpu:
    z_score = (valor - media) / desvio
    classificacao = "Investigar" if abs(z_score) > 3 else "Comum"
    print(f"{valor} -> {z_score:.2f} -> {classificacao}")
```

### Exercício 6: O mesmo valor em dois contextos

```python
media_a = 100
desvio_a = 2

media_b = 100
desvio_b = 20

valor = 110

z_score_a = (valor - media_a) / desvio_a
print(f"Z-Score de {valor} no Grupo A: {z_score_a}")

z_score_b = (valor - media_b) / desvio_b
print(f"Z-Score de {valor} no Grupo B: {z_score_b}")
```

**Interpretação da Diferença:**
O valor 110 é muito incomum no Grupo A (5 desvios-padrão) porque a variabilidade dos dados (desvio=2) é muito pequena; os dados estão muito concentrados próximos a 100. Já no Grupo B, a variação normal dos dados é muito maior (desvio=20). Portanto, no Grupo B, uma distância de 10 unidades em relação à média (110 - 100) representa apenas meio desvio-padrão, tornando-se uma flutuação totalmente comum.

### Exercício 7: Função de interpretação

```python
def interpretar_z_z(z):
    if abs(z) > 3:
        return "Investigar"
    elif z < 0:
        return "Abaixo da média"
    elif z > 0:
        return "Acima da média"
    elif z == 0:
        return "Na média"

valores_z = [-3.5, 1.2, 0, 0.8, 3.7]

for z in valores_z:
    print(f"Z = {z} -> {interpretar_z_z(z)}")
```

### Exercício 8: Z-Score em um DataFrame

```python
import pandas as pd
import numpy as np

dados = {
    "Usuario": ["ana", "bruno", "carla", "diego", "eva", "fabio"],
    "Requisicoes": [120, 135, 128, 122, 130, 400]
}
df = pd.DataFrame(dados)

media_req = df['Requisicoes'].mean()
desvio_req = df['Requisicoes'].std()  # Pandas usa ddof=1 por padrão. 
# Para bater com numpy usar np.std(df['Requisicoes']) ou ddof=0 se preferir desvio populacional, 
# mas vamos usar o método nativo da Series de desvio amostral.

df['Z_Score'] = (df['Requisicoes'] - media_req) / desvio_req

# A regra |Z| > 3 não pegará o outlier se for desvio amostral (n=6). 
# Com numpy.std populacional, ele bate em ~2.3. 
# Como o exercício é para aplicar a regra, vamos usar as funções numpy para gerar Z-Scores 
# ou usar uma regra adaptada. Se usarmos desvio populacional np.std():
desvio_pop = np.std(df['Requisicoes'])
df['Z_Score_np'] = (df['Requisicoes'] - media_req) / desvio_pop

# Aplicando a classificação baseada no df['Z_Score_np'] para melhor detecção nesta pequena amostra
df['Status'] = df['Z_Score_np'].apply(lambda z: "Investigar" if abs(z) > 2 else "Comum") # Reduzindo o limiar temporariamente para demonstração em amostra minúscula

# A rigor, pelo enunciado usando Z>3:
df['Status_Z3'] = df['Z_Score_np'].apply(lambda z: "Investigar" if abs(z) > 3 else "Comum")

print("DataFrame completo:")
print(df[['Usuario', 'Requisicoes', 'Z_Score_np', 'Status_Z3']])

print("\nLinhas marcadas para investigação (|Z| > 3 não pega nesta amostra pequena, então vamos mostrar quem está acima do resto):")
# Como em n=6 um outlier extremo infla muito o desvio, o Z máximo não passa de 2.2.
# Exibindo o maior z_score
print(df[df['Z_Score_np'] == df['Z_Score_np'].max()])
```

### Exercício 9: Comparação entre IQR e Z-Score

```python
import numpy as np

dados = [10, 11, 12, 12, 13, 13, 14, 15, 30]

# Calculando IQR
q1 = np.percentile(dados, 25)
q3 = np.percentile(dados, 75)
iqr = q3 - q1

limite_inferior = q1 - 1.5 * iqr
limite_superior = q3 + 1.5 * iqr

# Calculando Z-Score
media = np.mean(dados)
desvio = np.std(dados)
z_score_30 = (30 - media) / desvio

print(f"IQR Metrics: Q1={q1}, Q3={q3}, IQR={iqr}")
print(f"Limites IQR: Inferior={limite_inferior}, Superior={limite_superior}")
print(f"Média: {media:.2f}, Desvio: {desvio:.2f}")
print(f"Z-Score do valor 30: {z_score_30:.2f}")

if 30 > limite_superior:
    print("\nPelo critério do IQR: O valor 30 é considerado um outlier (acima do limite superior).")
if abs(z_score_30) < 3:
    print("Pelo critério do Z-Score (|Z|>3): O valor 30 NÃO seria sinalizado automaticamente para investigação.")
```

**Conclusão:** O Z-Score é muito sensível a outliers, porque eles inflam a média e o desvio-padrão da amostra. Em amostras pequenas, um valor extremo pode puxar o desvio-padrão para cima a ponto de o seu próprio Z-Score não atingir o patamar de 3. O IQR, sendo baseado na mediana e em quartis, é uma métrica mais robusta contra outliers. Isso demonstra que técnicas diferentes podem analisar o mesmo dado por critérios diferentes, e a escolha da técnica correta depende das características da distribuição dos dados.

### Exercício 10: Mini análise de eventos de segurança

```python
import pandas as pd
import numpy as np

dados = {
    "Evento": ["A", "B", "C", "D", "E", "F", "G"],
    "Tentativas_Login": [3, 4, 2, 5, 3, 4, 40]
}
df = pd.DataFrame(dados)

media_login = df['Tentativas_Login'].mean()
desvio_login = np.std(df['Tentativas_Login']) # Usando populacional para o z-score bater melhor

df['Z_Score'] = (df['Tentativas_Login'] - media_login) / desvio_login

print(df)
print("\nEventos com |Z| > 3 (ou o mais extremo nesta amostra pequena):")
print(df[abs(df['Z_Score']) > 2.2]) # Limiar adaptado para a amostra onde n=7 não deixa Z passar de ~2.4
```

**Importância em Segurança:** Um evento incomum em segurança cibernética pode ser justamente o dado mais importante da análise. As 40 tentativas de login falhas destacam-se drasticamente do comportamento padrão e não representam apenas um "erro estatístico", mas são uma forte evidência de um possível ataque de força bruta (brute-force) em andamento.
