### Exercício 1: Distância em passos de desvio-padrão

```python
media = 100
desvio = 5
valor = 115

distancia = valor - media
print(f"Distância entre o valor e a média: {distancia}")

desvios_na_distancia = distancia / desvio
print(f"Quantos desvios-padrão cabem nessa distância: {desvios_na_distancia}")

z_score = (valor - media) / desvio
print(f"Z-Score: {z_score}")
```

**Interpretação:** O valor 115 está 3 desvios-padrão acima da média do seu conjunto.

### Exercício 2: Acima ou abaixo da média?

```python
casos = [85, 100, 120]
media = 100
desvio = 10

for valor in casos:
    z_score = (valor - media) / desvio
    if z_score < 0:
        posicao = "abaixo da média"
    elif z_score == 0:
        posicao = "exatamente na média"
    else:
        posicao = "acima da média"
        
    print(f"Valor: {valor} | Z-Score: {z_score} | Posição: {posicao}")
```

**Interpretação dos Sinais:**
- **Sinal Positivo:** Indica que o valor é maior que a média.
- **Sinal Negativo:** Indica que o valor é menor que a média.
- **Zero:** Indica que o valor é exatamente igual à média.
